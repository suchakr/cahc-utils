````python
import math
import time
import torch

def bmx(device,sz=1200) :
    dtype = torch.float
    start_time = time.time()
    device = torch.device(device)
    # sz = 201

    x = torch.linspace(-1*math.pi, 1*math.pi, sz, device=device, dtype=dtype)
    y = torch.sin(x)

    a = torch.randn((), device=device, dtype=dtype)
    b = torch.randn((), device=device, dtype=dtype)
    c = torch.randn((), device=device, dtype=dtype)
    d = torch.randn((), device=device, dtype=dtype)
    # print( x.shape, y.shape, a.shape, b.shape, c.shape, d.shape)

    learning_rate = 1e-6
    prev_loss = None
    for t in range(18000):
        y_pred = a + b * x + c * x ** 2 + d * x ** 3
        loss = (y_pred - y).pow(2).sum().item()
        if prev_loss is not None and loss > prev_loss :
            print(f'Loss increased from {prev_loss} to {loss}')
            break
        prev_loss = loss
        # if t<10 and t % 1 == 0: print(device, t, loss)
        # if t<1000 and t % 100 == 0: print(device, t, loss)
        if t>=1000 and t % 1000 == 0: print(device, t, loss)

        grad_y_pred = 2.0 * (y_pred - y)
        grad_a = grad_y_pred.sum()
        grad_b = (grad_y_pred * x).sum()
        grad_c = (grad_y_pred * x ** 2).sum()
        grad_d = (grad_y_pred * x ** 3).sum()
        a -= learning_rate * grad_a
        b -= learning_rate * grad_b
        c -= learning_rate * grad_c
        d -= learning_rate * grad_d

    print( x.shape, y.shape, a.shape, b.shape, c.shape, d.shape)
    end_time = time.time()
    print(f'{device} {end_time - start_time:.5f}\nResult: y = {a} + {b} x + {c} x^2 + {d} x^3\n\n')

bmx("cpu", 1400)
bmx("mps", 1400)

````

````python
import torch
import time
import pandas as pd

def benchmark(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        ans = func(*args, **kwargs)
        end = time.time()
        elapsed  = end - start
        # print(f"{func.__name__:10s} {elapsed:.3f} seconds")
        return ans ,(func.__name__, elapsed)
    return wrapper

@benchmark
def alloc(sz=1024, device='cpu') :
    device = torch.device(device)
    x = torch.randn(sz, sz).to(device)
    y = torch.randn(sz, sz).to(device)
    return x, y

@benchmark
def sin(x) : return torch.sin(x) 
@benchmark
def poly(x) : return x**2 + x**3
@benchmark
def matmul(x, y) : return x.mm(y)
@benchmark
def sub(x, y) : return x - y

def test_dev(sz=1024, device='cpu') :
    bms = []
    # print(f"\nTesting with size {sz}^2 matrix on device {device}")
    (x, y), bm = alloc(sz, device) ; bms.append(bm)
    _ , bm = matmul(x, y) ; bms.append(bm)
    _ , bm = sub(x, y) ; bms.append(bm)
    _ , bm = sin(x) ; bms.append(bm)
    _ , bm = poly(x) ;  bms.append(bm)
    ans = pd.DataFrame( bms, columns=['func', 'time_secs']).assign(device = device)
    # display(ans)
    return ans

def tester(sz):
    bm1 = test_dev(sz, 'cpu')
    bm2 = test_dev(sz, 'mps')
    df = pd.concat([bm1, bm2])[['device', 'func', 'time_secs']].groupby(['device', 'func']).mean().unstack()#
    df.plot.bar(logy=True, figsize=(8, 3), title=f'Benchmarking CPU vs MPS - {sz}^2 matrix')
    display(df.style.format("{:.3f}").set_caption(f'Benchmarking CPU vs MPS - {sz}^2 matrix'))

tester(1024)
tester(1024*2)
tester(1024*20)
````
